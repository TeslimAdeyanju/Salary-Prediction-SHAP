### 2.1.3 [Assumptions and Limitations](#24-assumptions-and-limitations)

**Assumptions**
- `salary_in_usd` is assumed to be standardized across all countries regardless of local currency.
- The dataset is considered representative of the global ML job market for the year 2024.
- High-cardinality categorical features such as `job_title` can be effectively captured using embedding layers in a deep learning model.

**Limitations**
- Potential **regional bias** due to overrepresentation from countries like the USA and India.
- The dataset lacks **industry/sector information** (e.g., tech, finance, healthcare), which may impact salary variance.
- **Total compensation** may be underestimated due to missing data on bonuses, equity, or benefits.
- **Temporal limitation**: The dataset only covers the year 2024 and does not account for salary trends over time.

###  4.1.5 [Analysis of Categorical Embeddings](#42-analysis-of-categorical-embeddings)


This subsection addresses the challenge of handling categorical variables with many unique values (high-cardinality). Traditional one-hot encoding becomes impractical with variables like `job_title` and `company_location` that have hundreds of categories. We'll implement embedding layers that learn dense representations of these categorical variables.

```python
# Simplified Categorical Variables Analysis for Embeddings
import numpy as np
import pandas as pd

# Use df_features_combined as the data source
df_to_use = df_features_combined

# Get categorical columns
categorical_data = df_to_use.select_dtypes(include=['object']).columns.tolist()
print(f"Analyzing {len(categorical_data)} categorical variables...")

print("\n📊 CATEGORICAL EMBEDDING RECOMMENDATIONS")
print("=" * 40)

# Store embedding info for presentation
embedding_summary = []

# Analyze each categorical variable
for col in categorical_data:
    unique_vals = df_to_use[col].nunique()
    
    # Simple rule for embedding dimension
    embed_dim = min(50, max(1, unique_vals // 2))
    
    # Decide encoding type
    encoding_type = 'embedding' if unique_vals > 10 else 'one_hot'
    
    # Store information for summary table
    embedding_summary.append({
        'Variable': col,
        'Unique Values': unique_vals,
        'Encoding Type': encoding_type,
        'Embedding Dimension': embed_dim if encoding_type == 'embedding' else 'N/A'
    })

# Create and display summary table
summary_df = pd.DataFrame(embedding_summary)
print(summary_df.to_string(index=False))

# Calculate summary statistics
high_card_vars = sum(1 for item in embedding_summary if item['Encoding Type'] == 'embedding')
low_card_vars = sum(1 for item in embedding_summary if item['Encoding Type'] == 'one_hot')
total_embed_params = sum(item['Unique Values'] * item['Embedding Dimension'] 
                         for item in embedding_summary 
                         if item['Encoding Type'] == 'embedding')

print("\n📋 SUMMARY STATISTICS")
print("=" * 40)
print(f"• High-cardinality variables (embeddings): {high_card_vars}")
print(f"• Low-cardinality variables (one-hot): {low_card_vars}")
print(f"• Total embedding parameters: {total_embed_params:,}")

# Display top embedding variables by parameter count (if any)
if high_card_vars > 0:
    embed_vars = [item for item in embedding_summary if item['Encoding Type'] == 'embedding']
    embed_vars.sort(key=lambda x: x['Unique Values'] * x['Embedding Dimension'], reverse=True)
    
    print("\n🔝 TOP 3 VARIABLES BY EMBEDDING SIZE")
    print("=" * 40)
    for i, var in enumerate(embed_vars[:3], 1):
        params = var['Unique Values'] * var['Embedding Dimension']
        print(f"{i}. {var['Variable']}: {params:,} parameters ({var['Unique Values']} x {var['Embedding Dimension']})")
```

## 🛠️ 5.2 Neural Network Model Architecture

My neural network architecture:

- Uses a flexible structure with configurable hidden layers
- Includes BatchNorm for training stability
- Applies dropout for regularization
- Implements Kaiming initialization for better convergence
- Tracks parameter count for model complexity assessment


```python

# =============================================================================
# SECTION 5.2: Neural Network Model Architecture - DEBUGGED
# =============================================================================

class ImprovedNeuralNetwork(nn.Module):
    def __init__(self, input_size, hidden_sizes=[512, 256, 128, 64], dropout_rate=0.3):
        super(ImprovedNeuralNetwork, self).__init__()
        
        self.input_size = input_size
        self.hidden_sizes = hidden_sizes
        self.dropout_rate = dropout_rate
        
        # Create layers dynamically
        layers = []
        prev_size = input_size
        
        for hidden_size in hidden_sizes:
            layers.extend([
                nn.Linear(prev_size, hidden_size),
                nn.BatchNorm1d(hidden_size),
                nn.ReLU(),
                nn.Dropout(dropout_rate)
            ])
            prev_size = hidden_size
        
        # Output layer
        layers.append(nn.Linear(prev_size, 1))
        
        self.network = nn.Sequential(*layers)
        
        # Initialize weights
        self._initialize_weights()
    
    def _initialize_weights(self):
        """Initialize weights using Xavier/Glorot initialization"""
        for module in self.modules():
            if isinstance(module, nn.Linear):
                nn.init.xavier_uniform_(module.weight)
                nn.init.constant_(module.bias, 0)
    
    def forward(self, x):
        return self.network(x)

# Initialize model with error handling
try:
    input_size = X_train_tensor.shape[1]
    model = ImprovedNeuralNetwork(input_size=input_size).to(device)
    print(f"✅ Model initialized with input size: {input_size}")
    print(f"Model structure:\n{model}")
except Exception as e:
    print(f"❌ Error initializing model: {e}")
    raise
```

```python
# Improved NN architecture plot with better spacing, labels, legend, and exports

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import FancyArrowPatch, Circle, Rectangle, Patch
import matplotlib.patheffects as pe

def plot_neural_network(ax, layer_sizes, layer_names=None,
                        layer_spacing=2.8,
                        base_vertical_spacing=1.3,
                        neuron_radius=None,
                        connection_alpha=0.25,
                        connection_samples_left=4,
                        connection_samples_right=3,
                        show_hidden_extras=True,
                        colors=None):
    """
    Plot a neural network architecture diagram with improved readability.

    Parameters:
    - ax: matplotlib.axes.Axes to plot on
    - layer_sizes: list[int], number of neurons in each layer
    - layer_names: list[str] (optional), name for each layer
    - layer_spacing: float, horizontal spacing between layers
    - base_vertical_spacing: float, vertical spacing multiplier between neurons
    - neuron_radius: float or None, node radius (auto if None)
    - connection_alpha: float, transparency of connections
    - connection_samples_left/right: int, sampled connections to avoid clutter
    - show_hidden_extras: bool, annotate hidden layers with BN/Dropout hints
    - colors: dict with keys 'input', 'hidden', 'output', 'layer_box'
    """

    if colors is None:
        colors = {
            'input': '#FF9966',     # richer salmon
            'hidden': '#6FB1FC',    # stronger sky blue
            'output': '#20B2AA',    # light sea green
            'layer_box': '#EFEFEF'  # light gray for layer container
        }

    if layer_names is None:
        layer_names = [f"Layer {i+1}" for i in range(len(layer_sizes))]

    # Adaptive sizing based on network size
    max_neurons = max(layer_sizes) if layer_sizes else 1
    if neuron_radius is None:
        neuron_radius = 0.22 if max_neurons <= 12 else 0.18

    # Compute vertical spacing per layer (slightly adaptive)
    vertical_spacings = []
    for size in layer_sizes:
        # Increase spacing a bit as layers get denser
        spacing = base_vertical_spacing * (1.0 + 0.02 * max(0, size - 8))
        vertical_spacings.append(spacing)

    # Center each layer vertically; track half-height for labels/boxes
    y_half_extents = [(max(0, size - 1) * vs / 2.0) for size, vs in zip(layer_sizes, vertical_spacings)]
    x_positions = np.arange(len(layer_sizes)) * layer_spacing

    text_pe = [pe.withStroke(linewidth=2.5, foreground='white')]  # outline text for readability

    # Legend (call once later)
    legend_handles = [
        Patch(facecolor=colors['input'], edgecolor='black', label='Input Layer'),
        Patch(facecolor=colors['hidden'], edgecolor='black', label='Hidden Layer'),
        Patch(facecolor=colors['output'], edgecolor='black', label='Output Layer')
    ]

    # Draw layers and neurons
    for i, (size, x, name) in enumerate(zip(layer_sizes, x_positions, layer_names)):
        # Layer name slightly above layer box
        box_top = y_half_extents[i] + (vertical_spacings[i] * 0.9 + neuron_radius * 2.2)
        ax.text(x, box_top + 0.6, name, ha='center', va='center', fontsize=12, fontweight='bold',
                bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=3),
                path_effects=text_pe)

        # Layer container box (height contains all neurons)
        if size > 1:
            rect = Rectangle((x - (neuron_radius + 0.2),
                              -y_half_extents[i] - (neuron_radius + 0.2)),
                             width=(neuron_radius + 0.2) * 2,
                             height=(y_half_extents[i] * 2) + (neuron_radius + 0.2) * 2,
                             facecolor=colors['layer_box'], edgecolor='black', alpha=0.35)
            ax.add_patch(rect)

        # Per-layer detail label (units + activation)
        if i == 0:
            detail = f"{size} units"
        elif i == len(layer_sizes) - 1:
            detail = f"{size} unit • Linear"
        else:
            detail = f"{size} units • ReLU"
        ax.text(x, -y_half_extents[i] - (neuron_radius + 0.55), detail,
                fontsize=10, ha='center', va='top', color='#333',
                bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=2),
                path_effects=text_pe)

        # Extra hints for hidden layers
        if show_hidden_extras and 0 < i < len(layer_sizes) - 1:
            ax.text(x, -y_half_extents[i] - (neuron_radius + 1.05), "BatchNorm",
                    fontsize=9, ha='center', va='top', color='#4B0082',
                    bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=1),
                    path_effects=text_pe)
            ax.text(x, -y_half_extents[i] - (neuron_radius + 1.55), "Dropout",
                    fontsize=9, ha='center', va='top', color='#DC143C',
                    bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=1),
                    path_effects=text_pe)

        # Determine neuron color by layer type
        if i == 0:
            neuron_color = colors['input']
        elif i == len(layer_sizes) - 1:
            neuron_color = colors['output']
        else:
            neuron_color = colors['hidden']

        # Y positions for neurons
        y_range = np.linspace(-y_half_extents[i], y_half_extents[i], size) if size > 1 else [0.0]
        for y in y_range:
            circle = Circle((x, y), neuron_radius, facecolor=neuron_color, edgecolor='black', lw=1)
            ax.add_patch(circle)
            # Activation labels right of the neuron
            if i < len(layer_sizes) - 1:
                ax.text(x + neuron_radius + 0.15, y, "ReLU" if i < len(layer_sizes) - 1 else "Linear",
                        fontsize=8, ha='left', va='center',
                        bbox=dict(facecolor='white', alpha=0.85, edgecolor='none', pad=0.8),
                        path_effects=text_pe)

    # Connections between layers (sampled to avoid clutter)
    for i in range(len(layer_sizes) - 1):
        left_y = np.linspace(-y_half_extents[i], y_half_extents[i], layer_sizes[i]) if layer_sizes[i] > 1 else [0.0]
        right_y = np.linspace(-y_half_extents[i+1], y_half_extents[i+1], layer_sizes[i+1]) if layer_sizes[i+1] > 1 else [0.0]

        left_idx = np.linspace(0, len(left_y) - 1, min(connection_samples_left, len(left_y))).astype(int)
        right_idx = np.linspace(0, len(right_y) - 1, min(connection_samples_right, len(right_y))).astype(int)

        for li in left_idx:
            for ri in right_idx:
                arrow = FancyArrowPatch((x_positions[i] + neuron_radius * 0.2, left_y[li]),
                                        (x_positions[i+1] - neuron_radius * 0.2, right_y[ri]),
                                        arrowstyle='-|>',
                                        color='black',
                                        alpha=connection_alpha,
                                        lw=0.8,
                                        mutation_scale=10,
                                        connectionstyle='arc3,rad=0.18')
                ax.add_patch(arrow)

    # Axes limits and styling
    x_pad = neuron_radius + 1.0
    ax.set_xlim(-x_pad, x_positions[-1] + x_pad if len(x_positions) else x_pad)
    max_y = max(max(y_half_extents) + 2.6, 5.5) if y_half_extents else 5.5
    ax.set_ylim(-max_y, max_y)
    ax.axis('off')

    # Legend (upper left by default)
    ax.legend(handles=legend_handles, loc='upper left', framealpha=0.9,
              facecolor='white', edgecolor='#DDD', fontsize=10)

# ---------- Usage (same variables you already use) ----------

# Set high-quality defaults for export
plt.rcParams['figure.dpi'] = 160
plt.rcParams['savefig.dpi'] = 350

# Set up the figure with extra top/bottom space for title and embeddings info
fig = plt.figure(figsize=(16, 11), constrained_layout=False)
# Leave more room on top and bottom; we’ll place embeddings info at the figure level
plt.subplots_adjust(top=0.90, bottom=0.18, left=0.06, right=0.98)
ax = fig.add_subplot(111)

# Define your network architecture
input_size = 8
hidden_layers = [16, 8, 4]
output_size = 1
layer_sizes = [input_size] + hidden_layers + [output_size]
layer_names = [
    'Input\n(Preprocessed + Embeddings)',
    'Hidden Layer 1\n(Dense, ReLU)',
    'Hidden Layer 2\n(Dense, ReLU)',
    'Hidden Layer 3\n(Dense, ReLU)',
    'Output\n(Salary Prediction)'
]

# Create the diagram
plot_neural_network(
    ax,
    layer_sizes,
    layer_names=layer_names,
    layer_spacing=3.0,            # slightly increased horizontal spacing
    base_vertical_spacing=1.35,   # a touch more vertical spacing
    neuron_radius=None,           # auto-adjust
    connection_alpha=0.28,        # slightly more visible
    connection_samples_left=4,
    connection_samples_right=3,
    show_hidden_extras=True
)

# Titles
plt.suptitle('Neural Network Architecture for Salary Prediction', fontsize=18, y=0.975, fontweight='bold')
fig.text(0.5, 0.925, 'Deep Learning Model with Embedding Layers and Regularization',
         ha='center', fontsize=14, style='italic')

# Embeddings info box (figure-level so it never gets clipped)
embed_props = dict(boxstyle='round', facecolor='#E6E6FA', alpha=0.9, edgecolor='#555')
embedding_text = (
    "Embeddings:\n"
    "- job_title: 153 categories → 50 dim\n"
    "- employee_residence: 88 categories → 44 dim\n"
    "- company_location: 77 categories → 38 dim"
)
fig.text(0.5, 0.08, embedding_text, ha='center', va='center', fontsize=11, bbox=embed_props)

# Tighten without squashing the reserved spaces (top/bottom margin kept by rect)
plt.tight_layout(rect=[0.04, 0.18, 0.98, 0.90])

# Save as both high-DPI PNG and SVG (vector)
plt.savefig('neural_network_architecture_improved.png', dpi=350, bbox_inches='tight', pad_inches=0.6, facecolor='white')
plt.savefig('neural_network_architecture_improved.svg', bbox_inches='tight', pad_inches=0.6, facecolor='white')

plt.show()
```

## 🏋️ 5.4 Training Loop with Early Stopping

The training process includes:

- Encapsulated training loop function for better organization
- Progress bars for real-time monitoring
- Learning rate scheduling for adaptive learning
- Early stopping to prevent overfitting
- Model checkpointing to save the best performing model
- Gradient clipping to prevent exploding gradients

```python
# =============================================================================
# SECTION 5.4: Training Loop with Early Stopping - DEBUGGED
# =============================================================================

def train_model_safe(model, train_loader, val_loader, criterion, optimizer, scheduler, 
                     num_epochs, patience):
    """
    Enhanced training function with proper error handling and checkpoint management
    """
    best_val_loss = float('inf')
    epochs_no_improve = 0
    history = {'train_loss': [], 'val_loss': []}
    
    print("🚀 Starting training...")
    
    for epoch in range(num_epochs):
        try:
            # Training phase
            model.train()
            train_loss = 0.0
            train_batches = 0
            
            for batch_idx, (batch_x, batch_y) in enumerate(train_loader):
                optimizer.zero_grad()
                outputs = model(batch_x)
                loss = criterion(outputs, batch_y)
                loss.backward()
                
                # Gradient clipping for stability
                torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                
                optimizer.step()
                train_loss += loss.item()
                train_batches += 1
            
            avg_train_loss = train_loss / train_batches
            
            # Validation phase
            model.eval()
            val_loss = 0.0
            val_batches = 0
            
            with torch.no_grad():
                for batch_x, batch_y in val_loader:
                    outputs = model(batch_x)
                    loss = criterion(outputs, batch_y)
                    val_loss += loss.item()
                    val_batches += 1
            
            avg_val_loss = val_loss / val_batches
            
            # Update learning rate
            scheduler.step(avg_val_loss)
            
            # Save history
            history['train_loss'].append(avg_train_loss)
            history['val_loss'].append(avg_val_loss)
            
            # Print progress
            if (epoch + 1) % 10 == 0 or epoch == 0:
                print(f'Epoch [{epoch+1}/{num_epochs}] - '
                      f'Train Loss: {avg_train_loss:.6f}, '
                      f'Val Loss: {avg_val_loss:.6f}')
            
            # Check for improvement
            if avg_val_loss < best_val_loss:
                best_val_loss = avg_val_loss
                epochs_no_improve = 0
                
                # Save best model with error handling
                try:
                    checkpoint = {
                        'epoch': epoch,
                        'model_state_dict': model.state_dict(),
                        'optimizer_state_dict': optimizer.state_dict(),
                        'scheduler_state_dict': scheduler.state_dict(),
                        'best_val_loss': best_val_loss,
                        'train_loss': avg_train_loss
                    }
                    torch.save(checkpoint, os.path.join('model_checkpoints', 'best_model.pth'))
                    print(f"✅ New best model saved at epoch {epoch+1}")
                except Exception as e:
                    print(f"⚠️ Warning: Could not save checkpoint: {e}")
            else:
                epochs_no_improve += 1
            
            # Early stopping
            if epochs_no_improve >= patience:
                print(f'Early stopping triggered after {epoch+1} epochs')
                break
                
        except Exception as e:
            print(f"❌ Error during training at epoch {epoch+1}: {e}")
            break
    
    # Load best model if checkpoint exists
    checkpoint_path = os.path.join('model_checkpoints', 'best_model.pth')
    if os.path.exists(checkpoint_path):
        try:
            checkpoint = torch.load(checkpoint_path)
            model.load_state_dict(checkpoint['model_state_dict'])
            print("✅ Best model weights loaded successfully")
        except Exception as e:
            print(f"⚠️ Warning: Could not load best model: {e}")
            print("Using current model weights instead")
    else:
        print("⚠️ No checkpoint found, using current model weights")
    
    return model, history

# Train the model
trained_model, training_history = train_model_safe(
    model, train_loader, val_loader, criterion, optimizer, scheduler, num_epochs, patience
)
```

## 🏋️ 5.5 Model Evaluation and Performance Tracking

```python
# =============================================================================
# SECTION 5.5: Model Evaluation - DEBUGGED
# =============================================================================

def evaluate_model_safe(model, test_loader, criterion, target_scaler):
    """Evaluate model with error handling"""
    model.eval()
    test_loss = 0.0
    predictions = []
    actuals = []
    
    print("📊 Evaluating model...")
    
    try:
        with torch.no_grad():
            for batch_x, batch_y in test_loader:
                outputs = model(batch_x)
                loss = criterion(outputs, batch_y)
                test_loss += loss.item()
                
                # Store predictions and actuals
                predictions.extend(outputs.cpu().numpy())
                actuals.extend(batch_y.cpu().numpy())
        
        # Convert to numpy arrays
        predictions = np.array(predictions)
        actuals = np.array(actuals)
        
        # Inverse transform to original scale
        predictions_orig = target_scaler.inverse_transform(predictions.reshape(-1, 1)).flatten()
        actuals_orig = target_scaler.inverse_transform(actuals.reshape(-1, 1)).flatten()
        
        # Calculate metrics
        from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
        
        mae = mean_absolute_error(actuals_orig, predictions_orig)
        mse = mean_squared_error(actuals_orig, predictions_orig)
        rmse = np.sqrt(mse)
        r2 = r2_score(actuals_orig, predictions_orig)
        
        print("✅ Model Evaluation Results:")
        print(f"   Test Loss (scaled): {test_loss/len(test_loader):.6f}")
        print(f"   MAE: ${mae:,.2f}")
        print(f"   RMSE: ${rmse:,.2f}")
        print(f"   R²: {r2:.4f}")
        
        return predictions_orig, actuals_orig, {
            'mae': mae, 'rmse': rmse, 'r2': r2, 'test_loss': test_loss/len(test_loader)
        }
        
    except Exception as e:
        print(f"❌ Error during evaluation: {e}")
        return None, None, None

# Evaluate the model
y_pred, y_true, metrics = evaluate_model_safe(trained_model, test_loader, criterion, target_scaler)
```

```python
# Configure MLflow
# Suppress verbose MLflow output
import logging
logging.getLogger('mlflow').setLevel(logging.ERROR)

# Check if a tracking URI is already set, otherwise use local directory
if not os.environ.get("MLFLOW_TRACKING_URI"):
    mlflow_dir = "mlruns"
    os.makedirs(mlflow_dir, exist_ok=True)
    mlflow.set_tracking_uri(f"file:{os.path.abspath(mlflow_dir)}")

# Start a new MLflow experiment for model comparison
experiment_name = "Model_Comparison"
try:
    mlflow.create_experiment(experiment_name)
except:
    pass
mlflow.set_experiment(experiment_name)

# Set up the plot style
plt.style.use('fivethirtyeight')
sns.set_palette('Set2')
```

```python
# Helper Functions
def calculate_metrics(y_true, y_pred):
    """Calculate and format regression metrics"""
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    return {
        'RMSE': round(rmse, 2),
        'MAE': round(mae, 2),
        'R²': round(r2, 3)
    }

def create_model_signature(X, y):
    """Create model signature for MLflow"""
    from mlflow.models.signature import infer_signature
    return infer_signature(X, y)

def format_table(df):
    """Format a DataFrame as a nice table"""
    try:
        from tabulate import tabulate
        return tabulate(df.values.tolist(), headers=df.columns.tolist(), 
                       tablefmt="grid", floatfmt=".3f")
    except ImportError:
        return df.to_string(index=False)
```

### Train Linear Regression Model

```python
# 2A. Train Linear Regression Model
print("Training Linear Regression model...")

with mlflow.start_run(run_name="Linear_Regression"):
    mlflow.log_param("model_type", "Linear Regression")
    
    # Train model
    lr_model = LinearRegression()
    lr_model.fit(X_train_np, y_train_np)
    
    # Make predictions
    lr_preds = lr_model.predict(X_test_np)
    lr_metrics = calculate_metrics(y_test_np, lr_preds)
    all_metrics['Linear Regression'] = lr_metrics
    
    # Log metrics to MLflow
    for metric_name, metric_value in lr_metrics.items():
        mlflow.log_metric(metric_name, metric_value)
    
    # Log model with signature
    try:
        signature = create_model_signature(X_train_np, y_train_np)
        mlflow.sklearn.log_model(
            lr_model, 
            "model", 
            signature=signature,
            input_example=X_train_np[:5]
        )
    except:
        mlflow.sklearn.log_model(lr_model, "model")
    
    print(f"Linear Regression metrics: RMSE={lr_metrics['RMSE']:.2f}, R²={lr_metrics['R²']:.3f}")
```

### Train Random Forest Model

```python
# 2B. Train Random Forest Model
print("Training Random Forest model...")

with mlflow.start_run(run_name="Random_Forest"):
    mlflow.log_param("model_type", "Random Forest")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("random_state", 42)
    
    # Train model
    rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
    rf_model.fit(X_train_np, y_train_np)
    
    # Make predictions
    rf_preds = rf_model.predict(X_test_np)
    rf_metrics = calculate_metrics(y_test_np, rf_preds)
    all_metrics['Random Forest'] = rf_metrics
    
    # Log metrics to MLflow
    for metric_name, metric_value in rf_metrics.items():
        mlflow.log_metric(metric_name, metric_value)
    
    # Log model with signature
    try:
        signature = create_model_signature(X_train_np, y_train_np)
        mlflow.sklearn.log_model(
            rf_model, 
            "model", 
            signature=signature,
            input_example=X_train_np[:5]
        )
    except:
        mlflow.sklearn.log_model(rf_model, "model")
    
    print(f"Random Forest metrics: RMSE={rf_metrics['RMSE']:.2f}, R²={rf_metrics['R²']:.3f}")
```

### Train XGBoost Model

```python
# 2C. Train XGBoost Model
print("Training XGBoost model...")

with mlflow.start_run(run_name="XGBoost"):
    mlflow.log_param("model_type", "XGBoost")
    mlflow.log_param("n_estimators", 100)
    mlflow.log_param("learning_rate", 0.1)
    mlflow.log_param("random_state", 42)
    
    # Train model
    xgb_model = xgb.XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
    xgb_model.fit(X_train_np, y_train_np)
    
    # Make predictions
    xgb_preds = xgb_model.predict(X_test_np)
    xgb_metrics = calculate_metrics(y_test_np, xgb_preds)
    all_metrics['XGBoost'] = xgb_metrics
    
    # Log metrics to MLflow
    for metric_name, metric_value in xgb_metrics.items():
        mlflow.log_metric(metric_name, metric_value)
    
    # Log model with signature
    try:
        signature = create_model_signature(X_train_np, y_train_np)
        mlflow.xgboost.log_model(
            xgb_model, 
            "model", 
            signature=signature,
            input_example=X_train_np[:5]
        )
    except:
        mlflow.xgboost.log_model(xgb_model, "model")
    
    print(f"XGBoost metrics: RMSE={xgb_metrics['RMSE']:.2f}, R²={xgb_metrics['R²']:.3f}")
```

### Compile Results and Create Comparison Table

```python
# 3. Compile Results and Create Comparison Table
models = ['Linear Regression', 'Random Forest', 'XGBoost', 'Neural Network (FFNN)']
metrics_df = pd.DataFrame({
    'Model': models,
    'RMSE': [all_metrics['Linear Regression']['RMSE'], 
             all_metrics['Random Forest']['RMSE'], 
             all_metrics['XGBoost']['RMSE'], 
             all_metrics['Neural Network (FFNN)']['RMSE']],
    'MAE': [all_metrics['Linear Regression']['MAE'], 
            all_metrics['Random Forest']['MAE'], 
            all_metrics['XGBoost']['MAE'], 
            all_metrics['Neural Network (FFNN)']['MAE']],
    'R²': [all_metrics['Linear Regression']['R²'], 
           all_metrics['Random Forest']['R²'], 
           all_metrics['XGBoost']['R²'], 
           all_metrics['Neural Network (FFNN)']['R²']]
})

# Display the results with nice formatting
print("\n📊 MODEL PERFORMANCE COMPARISON")
print(format_table(metrics_df))
print()
```

```python
# 4. Create Visualization for Model Comparison
metrics_long = pd.melt(metrics_df, id_vars=['Model'], var_name='Metric', value_name='Value')

# Create subplots for each metric
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
metrics_to_plot = ['RMSE', 'MAE', 'R²']

for i, metric in enumerate(metrics_to_plot):
    metric_data = metrics_long[metrics_long['Metric'] == metric]
    
    # For R², higher is better, so we don't want to reverse the axis
    if metric == 'R²':
        ascending = False
    else:
        ascending = True
        
    # Sort values based on metric
    metric_data = metric_data.sort_values('Value', ascending=ascending)
    model_order = metric_data['Model'].tolist()
    
    # Create barplot with pre-defined order to avoid warnings
    ax = axes[i]
    sns.barplot(x='Model', y='Value', data=metric_data, order=model_order, ax=ax)
    ax.set_title(f'Model Comparison: {metric}', fontsize=14)
    ax.set_xlabel('')
    
    # Set x-tick positions explicitly to avoid warnings
    ax.set_xticks(range(len(model_order)))
    ax.set_xticklabels(model_order, rotation=45, ha='right')
    
    # Add value labels on top of each bar
    for j, bar in enumerate(ax.patches):
        value = bar.get_height()
        ax.text(
            bar.get_x() + bar.get_width()/2, 
            value + 0.01 * (ax.get_ylim()[1] - ax.get_ylim()[0]), 
            f'{value:.3f}',
            ha='center', va='bottom', fontsize=10
        )

plt.tight_layout()

# Save the comparison plot and log as artifact
comparison_plot_path = 'model_comparison_plot.png'
plt.savefig(comparison_plot_path, bbox_inches='tight', dpi=300)

# Log the comparison plot in a new MLflow run
with mlflow.start_run(run_name="Model_Comparison_Summary"):
    mlflow.log_artifact(comparison_plot_path, "comparison_plots")
    
    # Log the metrics table as a CSV
    metrics_csv_path = 'model_metrics_comparison.csv'
    metrics_df.to_csv(metrics_csv_path, index=False)
    mlflow.log_artifact(metrics_csv_path, "comparison_data")

plt.show()
```
